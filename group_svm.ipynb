```python
import pandas as pd
import time
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

df = pd.read_csv("data/DataCoSupplyChainDataset.csv", encoding="latin1")
print("Dataset loaded successfully")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

features = ["Days for shipment (scheduled)", "Order Item Quantity", "Order Item Discount", "Order Item Product Price"]
target = "Late_delivery_risk"

X = df[features].copy()
y = df[target].copy()
X = X.fillna(X.median())

# SVM does not scale well to very large datasets - subsample for feasibility
X_sample = X.sample(n=8000, random_state=42)
y_sample = y.loc[X_sample.index]

X_train, X_test, y_train, y_test = train_test_split(
    X_sample, y_sample, test_size=0.20, random_state=42, stratify=y_sample
)

print("\nFeatures used:", features)
print("Target variable:", target)
print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model = SVC(kernel="rbf", C=1.0, random_state=42)

start_time = time.time()
model.fit(X_train_scaled, y_train)
training_time = time.time() - start_time

print("SVM trained successfully")
print(f"Training time: {training_time:.4f} seconds")
```

```python
y_pred = model.predict(X_test_scaled)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("SVM PERFORMANCE")
print("----------------")
print(f"Accuracy : {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"Precision: {precision:.4f} ({precision*100:.2f}%)")
print(f"Recall   : {recall:.4f} ({recall*100:.2f}%)")
print(f"F1 Score : {f1:.4f} ({f1*100:.2f}%)")
print(f"Training Time: {training_time:.4f} seconds")
print(f"\nNote: Trained on a random 8,000-row sample (out of {len(X)} total rows) since SVM does not scale efficiently to very large datasets.")
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

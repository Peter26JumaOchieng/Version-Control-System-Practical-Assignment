```python
import pandas as pd
import time

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# Load the shared DataCo Supply Chain dataset
df = pd.read_csv(
    "data/DataCoSupplyChainDataset.csv",
    encoding="latin1"
)

print("Dataset loaded successfully")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# All three group members must use these same features
features = [
    "Days for shipment (scheduled)",
    "Order Item Quantity",
    "Order Item Discount",
    "Order Item Product Price"
]

# All three members use the same target
target = "Late_delivery_risk"

X = df[features].copy()
y = df[target].copy()

# Handle possible missing values
X = X.fillna(X.median())

# Same train/test split for all three models
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nFeatures used:")
print(features)

print("\nTarget variable:")
print(target)

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(
    max_depth=10,
    random_state=42
)

start_time = time.time()

model.fit(X_train, y_train)

training_time = time.time() - start_time

print("Decision Tree trained successfully")
print(f"Training time: {training_time:.4f} seconds")
```

```python
y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("DECISION TREE PERFORMANCE")
print("-------------------------")
print(f"Accuracy : {accuracy:.4f} ({accuracy * 100:.2f}%)")
print(f"Precision: {precision:.4f} ({precision * 100:.2f}%)")
print(f"Recall   : {recall:.4f} ({recall * 100:.2f}%)")
print(f"F1 Score : {f1:.4f} ({f1 * 100:.2f}%)")
print(f"Training Time: {training_time:.4f} seconds")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nRESULTS FOR GROUP COMPARISON")
print("Model: Decision Tree")
print(f"Accuracy: {accuracy * 100:.2f}%")
print(f"Precision: {precision * 100:.2f}%")
print(f"Recall: {recall * 100:.2f}%")
print(f"F1 Score: {f1 * 100:.2f}%")
print(f"Training Time: {training_time:.4f} seconds")
```
